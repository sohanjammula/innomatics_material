```python
data['gender'].value_counts()
```

```python
data['gender'] = data['gender'].apply(lambda x:1 if x=='Male' else 0)
data
```

```python
data['latitude_region'].unique()
```

```python
smoking_status_encoder = {'Never':0,'Former':1,'Current':2}
alcohol_consumption_encoder = {'Low-None':0,'Moderate':1,'Heavy':2}
exercise_level_encoder = {'Active':3, 'Light':1, 'Moderate':2, 'Sedentary':0}
diet_type_encoder= {'Vegetarian':0, 'Omnivore':1, 'Pescatarian':2, 'Vegan':3}
sun_exposure_encoder = {'High':2, 'Low':0, 'Moderate':1}
income_level_encoder = {'High':2, 'Low':0, 'Middle':1}
latitude_region_encoder = {'High':2, 'Low':0, 'Mid':1}

disease_encoder = {'Healthy':0, 'Rickets_Osteomalacia':1, 'Anemia':2, 'Night_Blindness':3,
       'Scurvy':4}
```

```python
data['smoking_status'] = data['smoking_status'].apply(lambda x:smoking_status_encoder[x])
data['alcohol_consumption'] = data['alcohol_consumption'].apply(lambda x:alcohol_consumption_encoder[x])
data['exercise_level'] = data['exercise_level'].apply(lambda x:exercise_level_encoder[x])
data['diet_type'] = data['diet_type'].apply(lambda x:diet_type_encoder[x])
data['sun_exposure'] = data['sun_exposure'].apply(lambda x:sun_exposure_encoder[x])
data['income_level'] = data['income_level'].apply(lambda x:income_level_encoder[x])
data['latitude_region'] = data['latitude_region'].apply(lambda x:latitude_region_encoder[x])

data['disease_diagnosis'] = data['disease_diagnosis'].apply(lambda x:disease_encoder[x])


data
```

```python
data.columns
```

```python
## train test split
from sklearn.model_selection import train_test_split

train_df,test_df = train_test_split(data,test_size=0.2)

features = ['age', 'gender', 'bmi', 'smoking_status', 'alcohol_consumption',
       'exercise_level', 'diet_type', 'sun_exposure', 'income_level',
       'latitude_region', 'vitamin_a_percent_rda', 'vitamin_c_percent_rda',
       'vitamin_d_percent_rda', 'vitamin_e_percent_rda',
       'vitamin_b12_percent_rda', 'folate_percent_rda', 'calcium_percent_rda',
       'iron_percent_rda', 'hemoglobin_g_dl', 'serum_vitamin_d_ng_ml',
       'serum_vitamin_b12_pg_ml', 'serum_folate_ng_ml', 'symptoms_count',
       'has_night_blindness', 'has_fatigue', 'has_bleeding_gums',
       'has_bone_pain', 'has_muscle_weakness', 'has_numbness_tingling',
       'has_memory_problems', 'has_pale_skin']

x_train = train_df[features]
y_train = train_df['disease_diagnosis']

x_test = test_df[features]
y_test = test_df['disease_diagnosis']
```

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
dt_model = DecisionTreeClassifier(max_depth=3)
```

```python
adaboost_model = AdaBoostClassifier(estimator = dt_model , n_estimators=10)
```

```python
adaboost_model.fit(x_train,y_train)
```

```python
adaboost_model.estimators_
```

```python
from sklearn.metrics import classification_report
```

```python
y_pred = adaboost_model.predict(x_train)
print(classification_report(y_train,y_pred))
```

```python
y_pred = adaboost_model.predict(x_test)
print(classification_report(y_test,y_pred))
```
