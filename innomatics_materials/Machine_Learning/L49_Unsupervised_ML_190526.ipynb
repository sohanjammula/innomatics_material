```python
education_encoder = {'Basic':0,'2n Cycle':1,'Graduation':2,'Master':3,'PhD':4}

data['Education']=data['Education'].apply(lambda x:education_encoder[x])

data
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

data['Marital_Status'] = data[['Marital_Status']].apply(le.fit_transform)
data
```

```python
data.columns
```

```python
# If we don't want previous marketing campaign data for current customer segmentation -> we can drop those columns
data.drop(['AcceptedCmp3',
       'AcceptedCmp4', 'AcceptedCmp5', 'AcceptedCmp1', 'AcceptedCmp2',
       'Complain', 'Response'],axis=1,inplace=True)
data
```

```python
## For PCA , data needs to scaled
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_data = scaler.fit_transform(data)
scaled_data
```

```python
scaled_data = pd.DataFrame(scaled_data,columns=data.columns)
scaled_data
```

## PCA

```python
from sklearn.decomposition import PCA
pca = PCA(n_components = 3)

pca_data = pca.fit_transform(scaled_data)
```

```python
pca_data = pd.DataFrame(pca_data,columns = ['PC1','PC2','PC3'])
pca_data
```

```python
import seaborn as sns
sns.scatterplot(x='PC1',y='PC3',data=pca_data)

## Scatterplots of Principle components should not have any patterns, relationship
```

## Clustering

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, n_init='auto')
```

```python
kmeans.fit(pca_data)
```

```python
kmeans.cluster_centers_
```

```python
y_pred = kmeans.predict(pca_data)
```

```python
y_pred
```

```python
values, counts = np.unique(y_pred, return_counts=True)

print("Values:", values)
print("Counts:", counts)
```

```python
## WCSS score
kmeans.inertia_
```

## Elbow Method

```python
wcss_values = []
k_values = []

for k in range(2,20):
  k_values.append(k)
  kmeans = KMeans(n_clusters = k,n_init='auto')
  kmeans.fit(pca_data)
  wcss_values.append(kmeans.inertia_)
```

```python
wcss_values
```

```python
import matplotlib.pyplot as plt

plt.plot(k_values,wcss_values)
plt.xlabel('K')
plt.ylabel('WCSS')
plt.xticks(k_values)
plt.show()
```

```python
# Based on Elbow method, K=7 is the optimal number of clusters
```

```python
kmeans = KMeans(n_clusters=7, n_init='auto')
kmeans.fit(pca_data)
```
