# Text Splitters Demonstration

This notebook demonstrates four types of text splitters used in LangChain for splitting documents into chunks:

1. **Character-based Splitter**: Simple character-based splitting
2. **Recursive Character Splitter**: Intelligently splits text by recursively trying different separators
3. **Markdown Text Splitter**: Specialized splitter for Markdown documents
4. **Semantic Chunker**: Embedding-based splitting that respects semantic boundaries

## Why Text Splitters?

Text splitters are essential in RAG (Retrieval-Augmented Generation) applications because:
- Large documents need to be broken into smaller chunks
- Chunks must fit within LLM context windows
- Proper splitting preserves semantic meaning
- Overlapping chunks maintain context between adjacent pieces


````python
# Import necessary libraries
from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    MarkdownHeaderTextSplitter
)

# Sample text for demonstration
sample_text = """
Artificial Intelligence (AI) has revolutionized the way we interact with technology. 
Machine learning, a subset of AI, enables computers to learn from data without explicit programming.

Deep learning uses neural networks with multiple layers to process complex patterns. 
Natural language processing (NLP) helps computers understand and generate human language.

Computer vision enables machines to interpret visual information from the world. 
These technologies work together to create intelligent systems that can perform tasks previously thought to be exclusive to humans.

Robotics combines AI with mechanical engineering to create autonomous machines. 
These robots can navigate environments, manipulate objects, and make decisions based on sensor data.

The future of AI holds promise for solving complex problems in healthcare, climate change, 
and scientific research. However, ethical considerations must be addressed to ensure 
responsible development and deployment of AI systems.
"""

print("Sample text length:", len(sample_text), "characters")
print("\nSample text preview:")
print(sample_text[:200] + "...")
````

## 1. Character-based Text Splitter

The `CharacterTextSplitter` is the simplest text splitter. It splits text into chunks based on a fixed character count.

**Characteristics:**
- Simple and straightforward
- Splits text at exact character positions
- May break words or sentences mid-way
- Good for simple use cases where structure doesn't matter much


````python
# Initialize Character Text Splitter
char_splitter = CharacterTextSplitter(
    separator= "\n",  # No separator - splits by character count only
    chunk_size=150,  # Maximum characters per chunk
    chunk_overlap=50,  # Maximum Overlap between chunks to maintain context
    length_function=len
)

# Split the text
char_chunks = char_splitter.split_text(sample_text)

print(f"Number of chunks created: {len(char_chunks)}")
print(f"\nChunk sizes: {[len(chunk) for chunk in char_chunks]}")
print("\n" + "="*60)
print("First 3 chunks:")
print("="*60)

for i, chunk in enumerate(char_chunks, 1):
    print(f"\n--- Chunk {i} ({len(chunk)} characters) ---")
    print(chunk)
    print()
````

````python
# Character splitter with a separator
char_splitter_with_separator = CharacterTextSplitter(
    separator="\n\n",  # Split on double newlines (paragraphs)
    chunk_size=100,
    chunk_overlap=10
)

char_chunks_sep = char_splitter_with_separator.split_text(sample_text)

print(f"Number of chunks with separator: {len(char_chunks_sep)}")
print(f"\nChunk sizes: {[len(chunk) for chunk in char_chunks_sep]}")
print("\n" + "="*60)
print("All chunks (with separator):")
print("="*60)

for i, chunk in enumerate(char_chunks_sep, 1):
    print(f"\n--- Chunk {i} ({len(chunk)} characters) ---")
    print(chunk)
    print()
````

## 2. Recursive Character Text Splitter

The `RecursiveCharacterTextSplitter` is more intelligent. It tries to split text using a hierarchy of separators, falling back to smaller separators if needed.

**Characteristics:**
- Tries multiple separators in order of preference
- Default separators: `["\n\n", "\n", " ", ""]`
- Preserves sentence/paragraph boundaries when possible
- Most commonly used splitter in practice
- Falls back to character-level splitting if needed


````python
# Initialize Recursive Character Text Splitter with default settings
recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=20,
    separators=["\n\n", "\n", ".", " ", ""]  # Order matters - tries from left to right
)

# Split the text
recursive_chunks = recursive_splitter.split_text(sample_text)

print(f"Number of chunks created: {len(recursive_chunks)}")
print(f"\nChunk sizes: {[len(chunk) for chunk in recursive_chunks]}")
print("\n" + "="*60)
print("All chunks (Recursive Character Splitter):")
print("="*60)

for i, chunk in enumerate(recursive_chunks, 1):
    print(f"\n--- Chunk {i} ({len(chunk)} characters) ---")
    print(chunk)
    print()
````

````python
# Recursive splitter with custom separators (sentence-first approach)
recursive_splitter_sentence = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=30,
    separators=[". ", "! ", "? ", "\n", " ", ""]  # Prioritize sentence endings
)

recursive_chunks_sentence = recursive_splitter_sentence.split_text(sample_text)

print(f"Number of chunks (sentence-first): {len(recursive_chunks_sentence)}")
print(f"\nChunk sizes: {[len(chunk) for chunk in recursive_chunks_sentence]}")
print("\n" + "="*60)
print("Chunks prioritizing sentence boundaries:")
print("="*60)

for i, chunk in enumerate(recursive_chunks_sentence, 1):
    print(f"\n--- Chunk {i} ({len(chunk)} characters) ---")
    print(chunk)
    print()
````

## 4. Semantic Chunker

The `SemanticChunker` (from `langchain_experimental`) takes a fundamentally different approach — instead of splitting by character counts or structural markers, it uses **embedding cosine similarity** to detect natural semantic boundaries.

**How it works:**
1. Splits text into individual sentences
2. Computes embeddings for each sentence
3. Measures cosine distance between adjacent sentence embeddings
4. Splits where the distance exceeds a configurable threshold

**Characteristics:**
- Chunks are driven by **meaning**, not fixed size
- Groups semantically related sentences together
- Produces more coherent chunks for downstream retrieval
- Requires an embeddings model (e.g., `OpenAIEmbeddings`)

**Breakpoint Threshold Types:**

| Type | Description |
|---|---|
| `percentile` *(default)* | Splits where distance > Nth percentile of all distances |
| `standard_deviation` | Splits where distance > mean + N × std dev |
| `interquartile` | Uses IQR-based outlier detection |
| `gradient` | Splits at points of steepest change in distance |

````python
import os
from dotenv import load_dotenv
from langchain_experimental.text_splitter import SemanticChunker
from langchain_huggingface import HuggingFaceEmbeddings

load_dotenv()


embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Initialize SemanticChunker using OpenAI Embeddings
# breakpoint_threshold_type="percentile" is the default:
#   splits wherever the cosine distance between adjacent sentences
#   exceeds the given percentile of all pairwise distances.
semantic_chunker = SemanticChunker(
    embeddings,
    breakpoint_threshold_type="percentile",
    breakpoint_threshold_amount=50,   # lower = more chunks, higher = fewer chunks
)

# create_documents() wraps each chunk in a LangChain Document object
semantic_chunks = semantic_chunker.create_documents([sample_text])

print(f"Number of semantic chunks: {len(semantic_chunks)}")
print(f"\nChunk sizes (chars): {[len(doc.page_content) for doc in semantic_chunks]}")
print("\n" + "="*60)
print("All Semantic Chunks (percentile method):")
print("="*60)

for i, doc in enumerate(semantic_chunks, 1):
    print(f"\n--- Chunk {i} ({len(doc.page_content)} characters) ---")
    print(doc.page_content)
    print()
````

## Key Parameters

### Common Parameters (CharacterTextSplitter & RecursiveCharacterTextSplitter):

1. **`chunk_size`**: Maximum size of chunks (in characters by default)
   - Too large: May exceed LLM context limits
   - Too small: May lose important context

2. **`chunk_overlap`**: Number of characters to overlap between chunks
   - Helps maintain context between adjacent chunks
   - Typically 10-20% of chunk_size

3. **`length_function`**: Function to measure chunk size (default: `len`)
   - Can use token counters for more accurate sizing

### RecursiveCharacterTextSplitter specific:

- **`separators`**: List of separators to try, in order of preference
  - Default: `["\n\n", "\n", " ", ""]`
  - Try larger structures first (paragraphs), then smaller ones

### MarkdownHeaderTextSplitter specific:

- Automatically handles Markdown structure
- Splits on headers, code blocks, lists, etc.
- No need to specify separators manually

### SemanticChunker specific:

- **`embeddings`**: Embeddings model used to compare sentence similarity (required)
- **`breakpoint_threshold_type`**: Strategy for detecting split points
  - `"percentile"` *(default)*: Splits where distance exceeds a given percentile
  - `"standard_deviation"`: Splits beyond mean + N × std dev
  - `"interquartile"`: Splits using IQR-based outlier detection
  - `"gradient"`: Splits at the steepest changes in cosine distance
- **`breakpoint_threshold_amount`**: Controls sensitivity (default `95` for percentile)
  - Lower value → more splits → smaller, more focused chunks
  - Higher value → fewer splits → larger, broader chunks
- **`min_chunk_size`**: Minimum number of sentences per chunk (prevents micro-chunks)


## Best Practices

1. **Choose the right splitter**:
   - Use `SemanticChunker` for highest retrieval quality when you have an embeddings model available
   - Use `RecursiveCharacterTextSplitter` for general documents when speed/cost matters
   - Use `MarkdownHeaderTextSplitter` for Markdown files
   - Use `CharacterTextSplitter` only for simple, uniform text

2. **Set appropriate chunk sizes**:
   - For character-based splitters: consider your LLM's context window, leave room for prompts and responses (typical: 500–2000 chars / 100–500 tokens)
   - For `SemanticChunker`: tune `breakpoint_threshold_amount` to control granularity

3. **Use overlap wisely** (character-based splitters):
   - 10-20% overlap is usually sufficient
   - Too much overlap wastes tokens; too little may lose context
   - `SemanticChunker` doesn't use overlap — semantic similarity naturally maintains coherence

4. **Test and iterate**:
   - Different content types may need different settings
   - Monitor retrieval quality and adjust accordingly

5. **Preserve metadata**:
   - Use `split_documents()` to maintain document metadata
   - Add chunk-specific metadata (chunk index, source, etc.)

## Splitter Comparison Summary

| Splitter | Speed | Cost | Coherence | Best For |
|---|---|---|---|---|
| `CharacterTextSplitter` | ⚡ Fast | Free | Low | Simple/uniform text |
| `RecursiveCharacterTextSplitter` | ⚡ Fast | Free | Medium | General documents |
| `MarkdownHeaderTextSplitter` | ⚡ Fast | Free | High (structure) | Markdown / docs |
| `SemanticChunker` | 🐢 Slow | Embedding cost | Highest | RAG pipelines where quality matters |
