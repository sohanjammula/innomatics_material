```python
from tensorflow.keras.applications import VGG16
```

```python
model = VGG16(weights="imagenet")
```

```python
#weights = None -- weights will be randomly init at the start of the traning
model = VGG16(weights = None, include_top=False, input_shape = (224,224,3))
```

```python
model.summary()
```

```python
from tensorflow.keras import Model #Functional implementation of keras
```

```python
output = model.output
```

```python
output
```

```python
from tensorflow.keras.layers import Conv2D, MaxPool2D, Flatten, Dense
```

```python
output
```

```python
output = Flatten()(output)
```

```python
output
```

```python
output = Dense(units =500,
                 kernel_initializer = "glorot_uniform",
                 bias_initializer = "glorot_uniform",
                 activation = "relu",
                 name = "fc_layer_1")(output)
```

```python
#output layer \

output = Dense(units =1,
                 kernel_initializer = "glorot_uniform",
                 bias_initializer = "glorot_uniform",
                 activation = "sigmoid",
                 name = "output_layer")(output)
```

```python
model.input
```

```python

#Functional implementation of Keras
custom_model = Model(inputs=model.input, outputs = output)
```

```python
custom_model
```

```python
from tensorflow.keras.optimizers import Adam
adam = Adam(learning_rate=0.001)
```

```python
custom_model.compile(loss = "binary_crossentropy", optimizer = adam, metrics = ["accuracy"])
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
train_gen = ImageDataGenerator(horizontal_flip = True,
                               rotation_range = 0.5,
                               shear_range = 0.3,
                               zoom_range=0.09,
                               rescale = 1/255.0)#minmaxscale
```

```python
test_gen = ImageDataGenerator(rescale = 1/255.0)
```

```python
train_data = train_gen.flow_from_directory(r"dataset/training_set",
                              target_size = (224, 224),
                              color_mode = "rgb",
                              class_mode = 'binary',
                              batch_size = 64,
                              shuffle=True)
```

```python
test_data = test_gen.flow_from_directory(r"dataset/test_set",
                              target_size =(224, 224),
                              color_mode = "rgb",
                              class_mode = 'binary',
                              batch_size = 64,
                              shuffle = False)
```

```python
custom_model.fit(train_data, epochs = 20)
```

```python
custom_model.save(r"vgg_cd.keras")
```
