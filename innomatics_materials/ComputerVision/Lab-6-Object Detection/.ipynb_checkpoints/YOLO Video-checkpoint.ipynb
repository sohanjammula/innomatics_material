```python
import cv2
from ultralytics import YOLO
import numpy as np
model = YOLO("yolov8n.pt")
```

```python
def class_wise_nms(boxes, classes, confidences, threshold=0.5):
    unique_classes = np.unique(classes)
    selected_boxes = []
    selected_classes = []
    selected_confidences = []
    boxes = np.array(boxes) 
    classes = np.array(classes) 
    confidences = np.array(confidences)



    for cls in unique_classes:
        class_indices = np.where(classes == cls)[0]
        class_boxes = boxes[class_indices]
        class_confidences = confidences[class_indices]

        while len(class_boxes) > 0:
            max_index = np.argmax(class_confidences)
            selected_boxes.append(class_boxes[max_index])
            selected_classes.append(cls)
            selected_confidences.append(class_confidences[max_index])
            current_box = class_boxes[max_index]

            iou = compute_iou(current_box, class_boxes)
            overlapping_indices = np.where(iou > threshold)[0]

            class_boxes = np.delete(class_boxes, overlapping_indices, axis=0)
            class_confidences = np.delete(class_confidences, overlapping_indices)

    return selected_boxes, selected_classes, selected_confidences

def compute_iou(box, boxes):
    intersect_x1 = np.maximum(box[0], boxes[:, 0])
    intersect_y1 = np.maximum(box[1], boxes[:, 1])
    intersect_x2 = np.minimum(box[2], boxes[:, 2])
    intersect_y2 = np.minimum(box[3], boxes[:, 3])
    intersect_area = np.maximum(0, intersect_x2 - intersect_x1 + 1) * np.maximum(0, intersect_y2 - intersect_y1 + 1)
    box_area = (box[2] - box[0] + 1) * (box[3] - box[1] + 1)
    boxes_area = (boxes[:, 2] - boxes[:, 0] + 1) * (boxes[:, 3] - boxes[:, 1] + 1)
    iou = intersect_area / (box_area + boxes_area - intersect_area)
    return iou
```

```python
def process_results(result, img_raw, conf_threshold):
    boxes = result.boxes.xyxy
    classes = result.boxes.cls
    confidences = result.boxes.conf
    class_dict = result.names
    final_boxes, final_classes, final_confidences = [], [], []
    for box, each_class, conf in zip(boxes,classes, confidences):
        x1, y1, x2, y2 = box.cpu().data.numpy().astype("int")
        #print(x1, y1, x2, y2)
        conf = conf.item()

        if conf > conf_threshold:
            each_class = int(each_class.item())
            class_name = class_dict[each_class]
            final_boxes.append([x1, y1, x2, y2])
            final_classes.append(each_class)
            final_confidences.append(conf)
    selected_boxes, selected_classes, selected_confidences = class_wise_nms(final_boxes, 
                                                                             final_classes, 
                                                                             final_confidences, 
                                                                             threshold=0.5)
    for box, each_class, conf in zip(selected_boxes, selected_classes, selected_confidences):
        x1, y1, x2, y2 = box
        #print(x1, y1, x2, y2)
        each_class = int(each_class.item())
        class_name = class_dict[each_class]
        cv2.rectangle(img_raw, (x1, y1), (x2, y2), (0,0,255), 2)
        cv2.putText(img_raw, class_name, (x1, y1-5), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255,0,0), 2)
    return img_raw
```

```python
def detect(frame):
    results = model.predict(source=frame, conf = 0, iou = 1)
    frame = process_results(results[0], frame, conf_threshold = 0.5)
    return frame
```

```python


cap = cv2.VideoCapture(r"D:\P1044014.mp4")
is_frame = True
while True:
    is_frame, frame = cap.read()
    if is_frame:
        annotated_frame = detect(frame)
        cv2.namedWindow("Video", cv2.WINDOW_GUI_EXPANDED)
        cv2.imshow("Video", annotated_frame)
        if cv2.waitKey(1) == ord("q"):
            break
    else:
        break

cap.release()
cv2.destroyAllWindows()
```
