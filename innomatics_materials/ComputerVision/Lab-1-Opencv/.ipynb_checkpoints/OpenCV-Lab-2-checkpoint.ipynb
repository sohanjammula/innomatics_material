# Part 1: Image Basics 📸

## What is OpenCV?
OpenCV (Open Source Computer Vision Library) is a powerful library for:
- **Image Processing**: Loading, displaying, manipulating images
- **Computer Vision**: Object detection, face recognition, tracking
- **Machine Learning**: Preprocessing data for ML models
- **Real-time Processing**: Working with cameras and video streams

## Key Concepts:
- **Images are arrays**: Images are stored as NumPy arrays
- **Color Channels**: BGR (Blue-Green-Red) vs RGB (Red-Green-Blue)
- **Image Dimensions**: Height × Width × Channels

### Step 1.3: Understanding Image Shape

**Image Shape Explained:**
- **Shape format**: (Height, Width, Channels)
- **Height**: Number of rows (pixels from top to bottom)
- **Width**: Number of columns (pixels from left to right)
- **Channels**: 
  - 3 for color images (BGR)
  - 1 for grayscale images
  - 4 for RGBA (with alpha channel)


```python
# Display image information
print("📊 Image Information:")
print(f"Shape: {image.shape}")
print(f"Height: {image.shape[0]} pixels")
print(f"Width: {image.shape[1]} pixels")
print(f"Channels: {image.shape[2]} channels")
print(f"Total pixels: {image.shape[0] * image.shape[1]}")
print(f"Memory size: {image.nbytes} bytes")
```

### Step 1.4: Display Images with Matplotlib

**Important Color Format Issue:**
- OpenCV loads images in **BGR** format
- Matplotlib expects **RGB** format
- We need to convert BGR → RGB for correct display


```python
# Convert BGR to RGB for matplotlib

#cv2.COLOR_BGR2RGB -- format of conversion
# Convert BGR to RGB for matplotlib
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Display both versions
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)  # This will show incorrect colors
plt.title("BGR Image (Incorrect Colors)")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(rgb_image)  # This will show correct colors
plt.title("RGB Image (Correct Colors)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

### Step 1.6: Region of Interest (ROI) Extraction

**What is ROI?**
- **ROI** = Region of Interest
- Extract a specific part of an image
- Useful for focusing on important areas
- **Syntax**: `image[y1:y2, x1:x2]` (rows first, then columns)


```python
#TO-DO

roi_image = rgb_image[y1:y2, x1:x2]

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(rgb_image)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(roi_image)
plt.title(f"ROI Extracted ({roi_image.shape[0]}x{roi_image.shape[1]})")
plt.axis('off')

plt.tight_layout()
plt.show()
```

### Step 1.9: Image Blurring Techniques

**What is Image Blurring?**
- **Purpose**: Reduce noise and smooth images
- **Effect**: Makes images less sharp, removes fine details
- **Applications**: Noise reduction, preprocessing for other operations, artistic effects

**Common Blur Types:**
1. **Gaussian Blur**: Natural-looking blur, weighted by distance
2. **Box Blur**: Simple average of neighboring pixels
3. **Median Blur**: Replaces pixel with median of neighborhood (great for salt-and-pepper noise)
4. **Bilateral Blur**: Preserves edges while smoothing


```python
#TO-DO

# Create a sample image with noise for blur demonstration
noisy_image = rgb_image.copy()

# Add some noise to make blur effects more visible
noise = np.random.randint(-30, 30, noisy_image.shape, dtype=np.int16)
noisy_image = np.clip(noisy_image.astype(np.int16) + noise, 0, 255).astype(np.uint8)

# Apply different blur techniques
gaussian_blur = cv2.GaussianBlur(noisy_image, (15, 15), 0)
box_blur = cv2.blur(noisy_image, (15, 15))
median_blur = cv2.medianBlur(noisy_image, 15)
bilateral_blur = cv2.bilateralFilter(noisy_image, 15, 80, 80)

plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(noisy_image)
plt.title("Original Image (with noise)")
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(gaussian_blur)
plt.title("Gaussian Blur")
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(box_blur)
plt.title("Box Blur")
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(median_blur)
plt.title("Median Blur")
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(bilateral_blur)
plt.title("Bilateral Blur")
plt.axis('off')

plt.tight_layout()
plt.show()
```

### Step 1.10: Understanding Blur Parameters

**Gaussian Blur Parameters:**
- `(15, 15)`: Kernel size (width, height) - must be odd numbers
- `0`: Standard deviation (0 = auto-calculate from kernel size)
- **Effect**: Larger kernel = more blur

**Box Blur Parameters:**
- `(15, 15)`: Kernel size
- **Effect**: Simple average of all pixels in kernel

**Median Blur Parameters:**
- `15`: Kernel size (single number for square kernel)
- **Effect**: Replaces each pixel with median value in neighborhood
- **Best for**: Salt-and-pepper noise removal

**Bilateral Blur Parameters:**
- `15`: Diameter of pixel neighborhood
- `80`: Color similarity threshold
- `80`: Space similarity threshold
- **Effect**: Smooths while preserving edges

### Step 1.9: Thresholding Techniques

**What is Thresholding?**
- **Purpose**: Convert grayscale to binary image
- **Process**: Set pixels above threshold to white, below to black
- **Types**: 
  - **Simple**: Fixed threshold value
  - **Adaptive**: Threshold varies by neighborhood
  - **Otsu**: Automatically finds optimal threshold

# Part 2: Video Processing 🎥

## What is Video Processing?
- **Video** = Sequence of images (frames) played rapidly
- **Frame Rate**: Frames per second (FPS)
- **Applications**: Real-time analysis, surveillance, video effects

## Key Concepts:
- **VideoCapture**: Read video files or camera streams
- **Frame-by-frame processing**: Process each image individually
- **Real-time processing**: Live camera feeds


### Step 2.1: Helper Function for Display

**Why Create a Helper Function?**
- **Consistency**: Same window size for all displays
- **Reusability**: Use the same function multiple times
- **Clean Code**: Avoid repeating window setup code


```python
def display_image(name, img):
    """
    Display image in a resized window
    
    Parameters:
    -----------
    name : str
        Window name
    img : numpy array
        Image to display
    """
    cv2.namedWindow(name, cv2.WINDOW_FREERATIO)
    cv2.resizeWindow(name, 500, 500)
    cv2.imshow(name, img)

print("✅ Display function created!")
```

### Step 2.2: Video Processing Basics

**Video Processing Workflow:**
1. **Open video**: `cv2.VideoCapture()`
2. **Read frames**: `cap.read()` returns (success, frame)
3. **Process frame**: Apply image processing techniques
4. **Display frame**: Show processed frame
5. **Check for exit**: Wait for 'q' key press
6. **Clean up**: Release video and close windows


```python
ord("q")
```

# Part 3: Morphological Operations 🔧

## What are Morphological Operations?
Morphological operations are image processing techniques that:
- **Process binary images** (black and white)
- **Use structuring elements** (kernels) to probe the image
- **Modify image structure** based on neighborhood pixels
- **Common applications**: Noise removal, shape analysis, feature extraction

## Key Operations:
1. **Erosion**: Shrinks objects
2. **Dilation**: Expands objects  
3. **Opening**: Erosion followed by dilation
4. **Closing**: Dilation followed by erosion


### Step 3.1: Create Sample Binary Image

**Why Binary Images?**
- Morphological operations work best on binary (black/white) images
- **Binary**: Only 0 (black) and 255 (white) values
- **Thresholding**: Convert grayscale to binary
- **Applications**: Document processing, medical imaging, object detection

### Step 3.2: Understanding Structuring Elements (Kernels)

**What is a Structuring Element?**
- **Kernel**: Small matrix used to probe the image
- **Shape**: Usually square, circle, or cross
- **Size**: Larger kernels = more aggressive operations
- **Purpose**: Defines neighborhood for morphological operations

### Step 3.3: Erosion Operation

**What is Erosion?**
- **Effect**: Shrinks white objects
- **Process**: If kernel fits completely inside white area → keep pixel white
- **Result**: Objects become smaller, holes become larger
- **Use cases**: Remove noise, separate connected objects

### Step 3.4: Dilation Operation

**What is Dilation?**
- **Effect**: Expands white objects
- **Process**: If kernel touches any white pixel → make center pixel white
- **Result**: Objects become larger, holes become smaller
- **Use cases**: Fill gaps, connect nearby objects

### Step 3.5: Opening Operation

**What is Opening?**
- **Definition**: Erosion followed by dilation
- **Effect**: Removes noise while preserving object size
- **Process**: 
  1. Erosion removes small noise
  2. Dilation restores object size
- **Use cases**: Noise removal, separating touching objects

### Step 3.6: Closing Operation

**What is Closing?**
- **Definition**: Dilation followed by erosion
- **Effect**: Fills gaps while preserving object size
- **Process**: 
  1. Dilation fills gaps and holes
  2. Erosion restores object size
- **Use cases**: Fill holes, connect broken objects
