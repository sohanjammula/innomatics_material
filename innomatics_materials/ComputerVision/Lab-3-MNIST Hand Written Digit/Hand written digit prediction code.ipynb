```python
import cv2
```

```python
import matplotlib.pyplot as plt
```

```python
from tensorflow.keras.models import load_model
```

```python
model = load_model("mnist_digit.keras")
```

```python
from tensorflow.keras.utils import normalize
```

```python
from PIL import Image
```

```python
import numpy as np
```

```python
def predict(path):
#     img = Image.open(path)
#     img = img.convert("L")
#     print(img.size)
#     img = img.resize((28,28))
#     display(img)
#     img = np.asarray(img)
    img = cv2.imread(path,0)# ,0 -- read in grayscale
    img = cv2.resize(img,(28,28))
    
    img = img.flatten()
    img = np.array([img])
    img = normalize(img)
    pred = model.predict(img,verbose = False)
    pred = np.argmax(pred, axis = 1)
    return pred
```

```python
predict(r"sample1_img.jpg")
```
