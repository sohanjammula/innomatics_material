```python
#tqdm -- to show the progress

#X = [Img_array1, Img_array2,Img_array3....Img_arrayn]
#y = ["2","5","3","4"]
X =[] #Input or features -- pixels values
y= [] #target -- labels

#crawling to all the folders
for each_folder in tqdm.tqdm(folder_paths):
    # get the classname from folder_path -- Get the folder name -- classname
    class_name = each_folder.split("/")[-1] #variable -y
    #print(each_folder+ "/*.jpg")
    #crawling to all the image in the each_folder
    for img_path in glob.glob(each_folder + "/*.jpg"):
        #print(img_path)
        #Read the image from the path
        img = cv2.imread(img_path,0)
            
        #flatten to covert 28,28 to 784 pixels to form a row
        img_array = img.flatten()
        #append the img in X
        X.append(img_array)
        #append the class_name
        y.append(class_name)
```

```python
X = np.array(X)
y = np.array(y)
```

### Split into train and test

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size = 0.8, random_state = 35)
```

```python
X_train
```

```python
X_test
```

```python
#Scaling

from tensorflow.keras.utils import normalize # Scale -- 0 to 1 from 0 to 255
```

```python
X_train = normalize(X_train.values.astype("uint"))
```

```python
X_test = normalize(X_test.values.astype("uint"))
```

```python
###Convertin the target to categorical##
```

```python
y
```

```python
from tensorflow.keras.utils import to_categorical #Convert the target into one hot representation
```

```python
y_train
```

```python
y_train = to_categorical(y_train)
```

```python
y_train
```

### Build the NN Architecture

```python
#Sequential
from tensorflow.keras.models import Sequential
```

```python
from tensorflow.keras.layers import Dense, Input
```

### Init the NN model

```python
nn_model = Sequential()
```

### Add the input layer

```python
X_train.shape[1]
```

```python
inp_layer = Input(shape = (X_train.shape[1],))
nn_model.add(inp_layer)
```

## Add Hidden Layer

```python
hidden_layer = Dense(kernel_initializer = "glorot_uniform", 
                     bias_initializer = "random_uniform", 
                     activation = "relu",
                     units = 200)
```

```python
nn_model.add(hidden_layer)
```

### Add Output Layer

```python
output_layer = Dense(kernel_initializer = "glorot_uniform", 
                     bias_initializer = "random_uniform", 
                     activation = "softmax",
                     units = 10)# classes -- 10 -- 0 to 9
```

```python
nn_model.add(output_layer)
```

```python
nn_model.summary()
```

### Compile the model

```python
nn_model.compile(loss = "categorical_crossentropy", optimizer =  "adam", metrics = ["accuracy"])
```

```python
from tensorflow.keras.utils import plot_model
```

```python
plot_model(nn_model)
#            show_shapes =  True, 
#        show_layer_names = True,
#           show_dtype = True,show_layer_activations = True, expand_nested  = True )
```

```python
from tensorflow.keras.callbacks import ModelCheckpoint
```

```python
ckpt = ModelCheckpoint("best_epoch_mnist.keras", 
                       monitor = "val_accuracy", 
                       save_best_only = True, 
                       mode = "max",
                      verbose = True)
```

```python
#batch_size -- No of observation to update the weights and send at once
#epochs -- no of iterations
train_logs = nn_model.fit(X_train, y_train, batch_size = 32, epochs = 10, 
                          validation_split = 0.1, callbacks = [ckpt])
```

```python
accuracy_list = train_logs.history["accuracy"]
```

```python
epochs = list(range(1,len(accuracy_list)+1))
```

```python
plt.plot(epochs, accuracy_list)
plt.title("Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.show()
```

```python
loss_list = train_logs.history["loss"]
```

```python
plt.plot(epochs, loss_list)
plt.title("Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()
```

```python
val_accuracy_list = train_logs.history["val_accuracy"]
```

```python
plt.plot(epochs, val_accuracy_list)
plt.title("Val Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Val Accuracy")
plt.show()
```

```python
val_loss = train_logs.history["val_loss"]
```

```python
plt.plot(epochs, val_loss)
plt.title("Val Loss")
plt.xlabel("Epochs")
plt.ylabel("Val Loss")
plt.show()
```

```python
def predict(path):
    img = cv2.imread(path,0)
    img = cv2.resize(path, (28,28))
    img = img.flatten()
    img = np.array([img])#1,784
    img = normalize(img)
    pred = nn_model.predict(img,verbose = False)
    pred = np.argmax(pred, axis = 1)
    return pred
```
