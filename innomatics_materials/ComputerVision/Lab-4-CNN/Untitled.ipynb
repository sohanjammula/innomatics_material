```python
def sample_function():
    l = []
    for i in range(10):
        l.append(i)
    return l
```

```python
sample_function()
```

```python
def sample_function():
   
    for i in range(10):
       yield i
   
```

```python
sample_function()
```

```python
for eac_ele in sample_function():
    print(eac_ele)
```
